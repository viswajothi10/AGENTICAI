```python
class Corpus(object):
    def __init__(self, path):
        self.filename = path
        self.nlp = spacy.blank("en")

    def __iter__(self):
        with open(self.filename, 'r', encoding='utf-8', errors='replace') as i:
            reader = csv.reader(i, delimiter=',')
            for _, abstract in reader:
                tokens = [t.text.lower() for t in self.nlp(abstract)]
                yield tokens

documents = list(Corpus(r"D:\AGENTICAI\arxiv.csv"))
documents[:5]
documents[:5]
```
