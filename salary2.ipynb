```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
ds=pd.read_csv('/content/drive/MyDrive/salary_dataset.csv')
```

```python
x=ds.drop(['Salary','Unnamed: 0'],axis=1)
Y=ds["Salary"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, Y, test_size=0.3, random_state=42)
```

```python
model = LinearRegression()
model=model.fit(x_train,y_train)
```

```python
predictSalary=model.predict(x_test)
```

```python
predictSalary
```

```python
errors=mean_squared_error(y_test,predictSalary)
print("Mean_absolute error:",errors)
r2=r2_score(y_test,predictSalary)
print("R2 score:",r2)

errors=mean_absolute_error(y_test,predictSalary)
print("Mean_absolute error:",errors)

mean_absolute_error= 0.07744778459829141

```
